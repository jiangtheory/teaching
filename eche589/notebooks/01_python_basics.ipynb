## 2. Variables and types

A variable gives a name to a stored value. Python determines the type from the value.

```python
sample_count = 4
temperature_c = 25.0
sample_name = "reactor outlet"
measurement_valid = True

print(sample_count, type(sample_count))
print(temperature_c, type(temperature_c))
print(sample_name, type(sample_name))
print(measurement_valid, type(measurement_valid))
```

The four types above are `int`, `float`, `str`, and `bool`. Use clear names that describe the physical meaning of each value.

```python
temperature_k = temperature_c + 273.15
print("Temperature in kelvin =", temperature_k)
```

## 3. Strings

A string is a sequence of characters enclosed by quotation marks.

```python
course = "ECHE 589"
topic = "Machine Learning"
message = course + " introduces " + topic

print(message)
print("Number of characters =", len(message))
print("First character =", message[0])
print("First four characters =", message[:4])

value = 0.3
print(f"This is f-string: {value:0.4f}")
```

Python begins counting positions at zero. The expression `message[0]` therefore returns the first character.

```python
print("Sample =", sample_name)
print("Temperature =", temperature_c, "degrees Celsius")
```

## 4. Logical expressions

A comparison produces either `True` or `False`.

```python
print("5 equals 4", 5 == 4)
print("2 is less than 3", 2 < 3)
print("5 is not equal to 4", 5 != 4)
print("3 is greater than or equal to 3", 3 >= 3)
```

Logical operators combine comparisons. Python uses `and`, `or`, and `not`.

```python
lower_limit = 20.0
upper_limit = 30.0
within_range = temperature_c >= lower_limit and temperature_c <= upper_limit

print("Temperature is within range =", within_range)
```

## 5. Lists

A list stores an ordered collection of values. Square brackets define a list.

```python
temperatures_c = [22.1, 23.5, 24.0, 25.2]

print("Complete list", temperatures_c)
print("Number of measurements", len(temperatures_c))
print("First measurement", temperatures_c[0])
print("Last measurement", temperatures_c[-1])
print("First two measurements", temperatures_c[:2])
```

Lists are mutable. Their contents can change after creation.

```python
temperatures_c.append(26.0)
temperatures_c[0] = 22.5
print(temperatures_c)
```

## 6. Tuples

A tuple is another ordered collection. Parentheses commonly define a tuple. A tuple cannot be changed after creation.

```python
acceptable_range_c = (20.0, 30.0)

print("Complete tuple", acceptable_range_c)
print("Lower limit", acceptable_range_c[0])
print("Upper limit", acceptable_range_c[1])
print("Number of values", len(acceptable_range_c))
```

## 7. Combine the ideas

This final example combines arithmetic, variables, strings, logical expressions, lists, and a tuple.

```python
run_name = "Run A"
measurements_k = [297.8, 298.2, 299.1, 298.7]
operating_range_k = (295.0, 300.0)

average_temperature_k = sum(measurements_k) / len(measurements_k)
within_operating_range = (
    average_temperature_k >= operating_range_k[0]
    and average_temperature_k <= operating_range_k[1]
)

print(run_name)
print("Average temperature =", round(average_temperature_k, 2), "K")
print("Average is within operating range =", within_operating_range)
```
